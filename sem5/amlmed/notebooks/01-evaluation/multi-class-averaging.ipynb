```python
import numpy as np
```

```python
import sklearn.metrics
```

```python
y_true = np.array([1, 2, 2, 1, 2, 3, 2, 3, 1, 1, 3, 2, 3, 2, 3, 1, 1, 3, 2, 3])
y_pred = np.array([1, 2, 1, 2, 2, 1, 2, 3, 1, 2, 3, 1, 2, 3, 3, 2, 1, 1, 2, 3])
```

```python
cm = sklearn.metrics.confusion_matrix(y_true=y_true, y_pred=y_pred)
cm
```

```python
weights = cm.sum(axis=1) / cm.sum()
weights
```

```python
precisions = sklearn.metrics.precision_score(y_true=y_true, y_pred=y_pred, average=None)
precisions
```

```python
[3/7, 4/8, 4/5]
```

```python
sklearn.metrics.precision_score(y_true=y_true, y_pred=y_pred, average="macro")
```

```python
precisions.mean()
```

```python
sklearn.metrics.precision_score(y_true=y_true, y_pred=y_pred, average="micro")
```

```python
(3 + 4 + 4) / (7 + 8 + 5)
```

```python
sklearn.metrics.precision_score(y_true=y_true, y_pred=y_pred, average="weighted")
```

```python
(precisions * weights).sum()
```

```python
recalls = sklearn.metrics.recall_score(y_true=y_true, y_pred=y_pred, average=None)
recalls
```

```python
[3/6, 4/7, 4/7]
```

```python
sklearn.metrics.recall_score(y_true=y_true, y_pred=y_pred, average="macro")
```

```python
recalls.mean()
```

```python
sklearn.metrics.recall_score(y_true=y_true, y_pred=y_pred, average="micro")
```

```python
(3 + 4 + 4) / (7 + 8 + 5)
```

```python
sklearn.metrics.recall_score(y_true=y_true, y_pred=y_pred, average="weighted")
```

```python
(recalls * weights).sum()
```

```python
sklearn.metrics.accuracy_score(y_true=y_true, y_pred=y_pred)
```
